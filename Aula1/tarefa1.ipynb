```python
import pandas as pd
df = pd.read_csv('drinks.txt')

# 1.Qual a média e desvio padrão do consumo de cerveja, licor, vinho e total de álcool?
colunas = ['beer_servings', 'spirit_servings', 'wine_servings', 'total_litres_of_pure_alcohol']
# Média
print('--- Média ---')
print(df[colunas].mean()) 

# Desvio padrão
print("")
print("--- Desvio padrão ---")
print(df[colunas].std())
```

```python
# 2.Qual o consumo de licor na França?
colunas = ['country', 'spirit_servings']
df[df['country'] == 'France'][colunas]
```

```python
# 3.Quais os 5 países nos quais se consome mais vinho?
colunas = ['country','wine_servings']
df.sort_values(by='wine_servings',ascending=False)[colunas].head(5)
```

```python
# 4.Quais os países nos quais o consumo de cerveja está acima da média?
media_cerveja = df['beer_servings'].mean()
print(media_cerveja)
colunas = ['country','beer_servings']
df[df['beer_servings'] > media_cerveja][colunas]
```

```python
# 5.Quais países nos quais o consumo total de álcool está acima do Brasil, em ordem do menor para o maior?
consumo_brasil = df[df['country'] == 'Brazil']['total_litres_of_pure_alcohol'].values[0]
print(consumo_brasil)
paises_acima = df[df['total_litres_of_pure_alcohol'] > consumo_brasil]
colunas = ['country','total_litres_of_pure_alcohol']
paises_acima[colunas].sort_values(by='total_litres_of_pure_alcohol',ascending=True)
```
