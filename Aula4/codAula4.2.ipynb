```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set(style="darkgrid")
```

```python
iris = sns.load_dataset("iris")
iris.head()
```

```python
# Distribuição conjunta através da combinação do histograma e do gráfico de dispersão -> joinplot()
sns.jointplot(x="sepal_length", y="petal_length", kind = 'reg', data=iris)
plt.show()
```

```python
# Gráfico de hexbin 
sns.jointplot(x="sepal_length", y="petal_length", kind = 'hex', color = 'black', data=iris) 
plt.show()
```

```python
# Gráfico de densidade
sns.jointplot(x="sepal_length", y="petal_length", kind = 'kde', data=iris)
plt.show()
```

```python
sns.pairplot(iris)
plt.show()
```

```python
sns.pairplot(iris, kind = 'reg')
plt.show()
```

```python
sns.pairplot(iris, hue='species')
plt.show()
```
