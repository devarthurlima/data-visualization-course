```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

df = sns.load_dataset("titanic")
df.head()
```

```python
# Crie uma visualização que mostre a distribuição do valor da passagem (“fare”), de acordo com a cidade na qual o passageiro embarcou ("embark_town"). Você pode escolher a técnica que julgar mais adequada;
plt.figure(figsize=(10,5))
plt.title('Distribuição do Valor da Passagem por Cidade de Embarque')
sns.boxplot(x='embark_town', y='fare', data=df)
plt.show()
```

```python
# Crie uma visualização que mostre as distribuições do valor da passagem, de acordo com a cidade de embarque, sexo e se o passageiro sobreviveu ou não. Você pode escolher a técnica que julgar mais adequada. Que diferença você consegue notar nas distribuições dos passageiros que sobreviveram em relação aos que não sobreviveram?
sns.catplot(x="embark_town", y="fare", hue = 'sex', col = 'survived', kind = 'violin', split = True, data=df)
plt.show()
print("Que diferença você consegue notar nas distribuições dos passageiros que sobreviveram em relação aos que não sobreviveram? Passageiros com tarifas mais altas tiveram chances significativamente maiores de sobreviver, enquanto os não sobreviventes se concentraram em tarifas baixas. As mulheres tiveram taxa de sobrevivência muito superior à dos homens em qualquer faixa de preço. A cidade de Cherbourg teve as tarifas mais altas e maior índice de sobreviventes de alta renda; Queenstown concentrou tarifas baixas e pouca variação de sobrevida")
```
