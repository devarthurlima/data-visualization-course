```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('houses_to_rent_v3.txt')
df.head()
```

```python
# describe() -> realiza diversos cálculos estatísticos que informam sobre a distribuição dos dados
df['area'].describe()
# count é a contagem dos dados
# mean é a média aritmética
# std é o desvio padrão
```

```python
# A biblioteca matplotlib permite criar e modificar histogramas facilmente através da função hist()
plt.figure(figsize=(10,5))
plt.style.use
plt.title('Distribuição da área dos imóveis')
plt.hist(df["area"])
plt.show()
```

```python
# Aumentando a quantidade de bins
plt.figure(figsize=(10,5))
plt.title("Distribuição de áreas do imóveis")
plt.hist(df['area'], bins=25)
plt.show
```

```python
# Gráficos de densidade com a biblioteca seaborn -> distplot()
plt.figure(figsize=(10, 5))
plt.title('Distribuição de áreas dos imóveis')
sns.distplot(df['area'])
plt.show()
```

```python
# Se você quiser apenas o gráfico de densidade, pode-se usar a função kdeplot()
plt.figure(figsize=(10,5))
plt.title("Distribuição de áreas do imóveis")
sns.kdeplot(df['area'], shade=True)
plt.show()
```
