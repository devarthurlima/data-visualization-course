```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

df = sns.load_dataset("mpg")
df.head()
```

```python
# Crie uma visualização que mostre a relação entre a aceleração (‘acceleration’) e o consumo de combustível (‘mpg’) dos veículos, em que seja possível diferenciar o país de origem (‘origin’). Sabendo disso, existe uma correlação entre a aceleração e o consumo? De que tipo? A correlação é a mesma para os diferentes países de origem? Você pode escolher a técnica que julgar mais adequada!
plt.figure(figsize=(8,5))
plt.title("Relação entre a aceleração e o consumo de combustível diferenciando por país")
sns.scatterplot(x="acceleration", y="mpg", hue="origin", data=df)
plt.show()

print("Existe uma correlação entre a aceleração e o consumo? Sim.") 
print("De que tipo? Correlação alta moderada") 
print("A correlação é a mesma para os diferentes países de origem? Não exatamente, o EUA uma dispersão mais ampla")
```

```python
# Crie uma visualização que mostre a relação entre a aceleração (‘acceleration’) e o consumo de combustível (‘mpg’) dos veículos, em que seja possível visualizar o ano do modelo (model_year). Existe alguma relação entre as três variáveis?
plt.figure(figsize=(8,5))
plt.title("Relação entre a aceleração e o consumo de combustível diferenciando por ano do modelo")
sns.scatterplot(x="acceleration", y="mpg", hue="model_year", data=df)
plt.show()

print("Existe alguma relação entre as três variáveis? Sim, quanto maior o ano de modelo, maior tende a ser o consumo de combustível, logo, eles possuem uma correlação alta positiva, enquanto a aceleração e o consumo continuam possuindo uma correlação alta moderada, a aceleração e o ano de modelo não possuem uma correlação aparente.")
```
