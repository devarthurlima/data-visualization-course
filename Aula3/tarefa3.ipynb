```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

df = pd.read_csv('dados.txt')
df.head()
```

```python
# Para o segundo e último passo, crie uma visualização que mostre a série temporal das vendas ao longo dos meses. Você pode, por exemplo, se perguntar se existe alguma tendência ou alguma periodicidade, OK?
plt.figure(figsize=(20,5))
plt.title('Vendas ao longo dos meses')
sns.lineplot(x="Month", y="Sales", marker="o" ,data=df)
plt.show()

print("Existe alguma tendência ou alguma periodicidade? As vendas mostram uma clara tendência ascendente ao longo do tempo.")
```
