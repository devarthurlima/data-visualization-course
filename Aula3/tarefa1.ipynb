```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

df = pd.read_csv('houses_to_rent_v3.txt')
```

```python
# Crie uma visualização que mostre a distribuição dos valores de aluguel, ‘rent amount (R$)’, dos imóveis. Pode utilizar o gráfico que você julgar mais adequado. Qual faixa de valores de aluguel é mais comum?
plt.figure(figsize=(10, 5))
plt.title('Distribuição de valores de aluguel')
plt.hist(df['rent amount (R$)'], bins=30)
plt.show()

print("Qual faixa de valores de aluguel é mais comum? É a faixa entre R$ 1.000 e R$ 3.000 com pico em torno de R$ 2.000 a R$ 2.500.")
```

```python
# Crie uma visualização que mostre a distribuição dos valores de seguro contra incêndio, 'fire insurance (R$)', dos imóveis. Existem valores incomuns?
plt.figure(figsize=(10, 5))
plt.title('Distribuição de valores de seguro contra incêndio')
sns.histplot(df['fire insurance (R$)'], kde=True)
plt.show()

print("Existem valores incomuns? Sim, existem valores incomuns acima de R$ 150 - R$ 200, chegando a ultrapassar R$ 300, enquanto a grande maioria dos imóveis tem seguro concentrado abaixo de R$ 50.")
```
