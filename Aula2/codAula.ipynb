```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

```

```python
# Criando meu primeiro gráfico de barras 
categorias = ['A','B','C','D']
valores = [10,20,15,5]
plt.bar(categorias,valores)
plt.xlabel('Grupos')
plt.ylabel("Tamanho")
plt.title("Meu primeiro gráfico de barras")
plt.show()
```

```python
# Alterando a cor e o tamanho 
plt.figure(figsize=(12,3))
plt.bar(categorias,valores,color='Darkred')
plt.xlabel('Grupos')
plt.ylabel('Tamanho')
plt.title("Meu primeiro gráfico de barras")
plt.show()
```

```python
# Tabela com dados estruturados de filmes 
filmes_uri = 'https://raw.githubusercontent.com/emanueles/datavis-course/master/assets/files/observable/movies.json'
filmes = pd.read_json(filmes_uri)
filmes.head()

# método describe() para obter medidas estatísticas sobre os atributos quantitativos
filmes.describe()
```

```python
# groupby para agrupar o dados e count para contar as quantidades
contagens = filmes.groupby(by='Genre').count()
contagens
```

```python
# .bar é o responsável pela criação das barras verticais 
plt.bar(contagens.index, contagens['Worldwide_Gross_M'])
plt.show()
```

```python
# countplot serve para gerar gráficos de barras de contagem de elementos
sns.set(style="darkgrid")
sns.countplot(x="Genre", data=filmes)
plt.show()
```

```python
# Acidionando titulo e alterando o tamanho e a cor do gráfico
sns.set(style="darkgrid")
plt.figure(figsize=(10,5))
plt.title('Quantidade de filmes por gênero')
sns.countplot(x="Genre", color = 'r', data=filmes)
plt.show()
```

```python
# Transformando em barras horizontais
plt.title('Quantidade de filmes por gênero')
sns.countplot(y="Genre", color = 'r', data=filmes)
plt.show()
```

```python
# barplot() mostra medidas associadas a esses grupos(média aritmética)
plt.figure(figsize=(10,5))
plt.title('Bilheteria média por gênero')
sns.barplot(x="Genre", y="Worldwide_Gross_M", color = 'b', ci = 'sd', data=filmes)
plt.show()
```

```python
# Adicionando a função sum(média) no estimator
plt.figure(figsize=(10,5))
plt.title('Bilheteria total por gênero')
sns.barplot(x="Genre", y="Worldwide_Gross_M", color = 'orange', estimator = sum, ci = None, data=filmes)
plt.show()
```

```python
# Gráfico de barras compostas
plt.figure(figsize=(15,8))
plt.title('Filmes por gênero em cada ano')
sns.countplot(x="Year", hue= 'Genre', data=filmes)
plt.show()
```

```python
# Trocando o countplot pelo barplot
plt.figure(figsize=(15,8))
plt.title('Média de bilheteria por gênero em cada ano')
sns.barplot(x="Year", y = 'Worldwide_Gross_M' , ci = None, hue= 'Genre', data=filmes)
plt.show()
```

```python
# Adicionando a paleta pastel e o atributo edgecolor que configura a cor do contorno das barras(0 a 1)
plt.figure(figsize=(15,8))
plt.title('Média de bilheteria por gênero em cada ano')
sns.countplot(x="Year", hue= 'Genre', palette = "pastel", edgecolor="0.5", data=filmes)
plt.show()
```

```python
# Comando para salvar o gráfico na pastas 
plt.figure(figsize=(15,8))
plt.title('Média de bilheteria por gênero em cada ano')
sns.countplot(x="Year", hue= 'Genre', palette = "pastel", edgecolor="0", data=filmes)
#plt.savefig('Média de bilheteria por gênero em cada ano.png')
```
