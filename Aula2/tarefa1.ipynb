```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

df = pd.read_json('dados.txt')
```

```python
# 2.Crie uma visualização que mostre o orçamento de todos os filmes do conjunto de dados. Qual o filme com maior orçamento e com menor orçamento?

plt.figure(figsize=(10, 12))
plt.barh(df['Film'], df['Budget_M'])
plt.show()

maior_orcamento = df[df['Budget_M'] == df['Budget_M'].max()]['Film'].values[0]
menor_orcamento = df[df['Budget_M'] == df['Budget_M'].min()]['Film'].values[0]

print("Maior Orçamento:", maior_orcamento, f"({df['Budget_M'].max()})")
print("Menor Orçamento:", menor_orcamento, f"({df['Budget_M'].min()})")
```

```python
# 3.Crie uma visualização que mostre a bilheteria de todos os filmes do conjunto de dados. Qual o filme com maior bilheteria e com menor bilheteria?


plt.figure(figsize=(10, 12))
plt.barh(df['Film'], df['Worldwide_Gross_M'])
plt.show()

menor_bilheteria = df[df['Worldwide_Gross_M'] == df['Worldwide_Gross_M'].min()]['Film'].values[0]
maior_bilheteria = df[df['Worldwide_Gross_M'] == df['Worldwide_Gross_M'].max()]['Film'].values[0]

print("Maior Bilheteria:", maior_bilheteria, f"({df['Worldwide_Gross_M'].max()})")
print("Menor Bilheteria:", menor_bilheteria, f"({df['Worldwide_Gross_M'].min()})")
```

```python
# 4.Crie uma visualização que mostre a quantidade de filmes por gênero em cada ano. Em que anos foram feitos menos filmes de ação? E em que ano foram feitos menos filmes de animação?

plt.figure(figsize=(8, 10))
plt.title('Quantidade de filmes por gênero em cada ano')
sns.countplot(y="Genre", hue="Year" , data=df)
plt.show()

print("Anos com menos filmes de ação: 2008 e 2010 (2 filmes cada)")
print("Anos com menos filmes de animação: 2008 e 2011 (1 filme cada)")
```

```python
# 5.Crie uma visualização que mostre a média de orçamento de filmes por gênero em cada ano. Em que ano os filmes de ação tiveram maior média de orçamento? E em que ano os filmes de animação tiveram a maior média de orçamento?

plt.figure(figsize=(15,8))
plt.title('Média de orçamento de filmes por gênero em cada ano')
sns.barplot(x="Year", y = 'Budget_M' , errorbar = None, hue= 'Genre', data=df)
plt.show()

print("Ano com maior média de orçamento em filmes de ação: 2007")
print("Ano com maior média de orçamento em filmes de animação: 2008")
```
