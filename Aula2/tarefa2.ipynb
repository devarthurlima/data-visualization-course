```python
# Importe as bibliotecas pandas e matplotlib, usando a palavra mágica inline para exibir as visualizações no notebook; e carregue os dados, que estão disponibilizados neste link:dados, em um DataFrame. Assim, você pode carregar os dados pela URI ou salvar em sua máquina e carregá-los localmente;
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
games = pd.read_csv("vgsales.csv")
```

```python
# Crie uma visualização que mostre a proporção de vendas global entre gêneros de jogos da empresa Nintendo;
nintendo_games = games[games['Publisher'] == 'Nintendo']

vendas_genero = nintendo_games.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False)

porcentagens = (vendas_genero / vendas_genero.sum()) * 100

plt.style.use('ggplot')
plt.figure(figsize=(15,5))
plt.bar(porcentagens.index, porcentagens)
plt.title('Proporção de vendas globais por gêneros de jogos da Nintendo')
plt.show()
```

```python
# Crie uma visualização que mostre a proporção de quantidade de jogos por plataforma a partir do ano de 2015.
jogos_2015 = games[games['Year'] >= 2015]

proporcao_plataforma = jogos_2015['Platform'].value_counts(normalize=True) * 100

plt.figure(figsize=(12, 5))
plt.bar(proporcao_plataforma.index, proporcao_plataforma)
plt.title('Proporção de Lançamento de Jogos por Plataforma (>= 2015)')
plt.show()
```
